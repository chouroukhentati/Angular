```python
import numpy as np
import torch
from torch.utils.data import Dataset
from pathlib import Path
from PIL import Image
import torchvision.transforms as T

class TeethSegDataset(Dataset):
    def __init__(self, images_dir, masks_dir, file_names, image_size=572, labelmap_data=None):
        self.images_dir = Path(images_dir)
        self.masks_dir = Path(masks_dir)
        self.file_names = file_names
        self.image_size = image_size

        self.image_tf = T.Compose([
            T.Resize((image_size, image_size)),
            T.ToTensor(),
        ])

        self.mask_resize = T.Resize(
            (image_size, image_size),
            interpolation=T.InterpolationMode.NEAREST
        )

        self.rgb_to_class = None
        self.class_names = None

        if labelmap_data is not None:
            self.rgb_to_class, self.class_names = self._create_rgb_to_class_map(labelmap_data)

    def _create_rgb_to_class_map(self, labelmap_data):
        rgb_to_class = {}
        class_names = []

        for idx, line in enumerate(labelmap_data):
            line = line.strip()
            if not line:
                continue

            left, right = line.split(":", 1)
            class_name = left.strip()
            rgb = tuple(int(x.strip()) for x in right.split(","))

            rgb_to_class[rgb] = idx
            class_names.append(class_name)

        return rgb_to_class, class_names

    def _rgb_mask_to_class_ids(self, mask):
        mask = mask.convert("RGB")
        mask_np = np.array(mask)
        class_mask = np.zeros(mask_np.shape[:2], dtype=np.int64)

        for rgb, class_id in self.rgb_to_class.items():
            matches = np.all(mask_np == rgb, axis=-1)
            class_mask[matches] = class_id

        return torch.from_numpy(class_mask).long()

    def __len__(self):
        return len(self.file_names)

    def __getitem__(self, idx):
        name = self.file_names[idx]

        img = Image.open(self.images_dir / name).convert("RGB")
        mask = Image.open(self.masks_dir / name)

        img = self.image_tf(img)
        mask = self.mask_resize(mask)

        if self.rgb_to_class is not None:
            mask = self._rgb_mask_to_class_ids(mask)
        else:
            mask = torch.as_tensor(np.array(mask), dtype=torch.long)
            if mask.ndim == 3:
                mask = mask[..., 0]

        return img, mask
```

```python
class DepthwiseSeparableConv(nn.Module):
    def __init__(self, in_channels, out_channels, dropout=0.0):
        super().__init__()
        layers = [
            nn.Conv2d(in_channels, in_channels, kernel_size=3, padding=1, groups=in_channels, bias=False),
            nn.BatchNorm2d(in_channels),
            nn.ReLU(inplace=True),
            nn.Conv2d(in_channels, out_channels, kernel_size=1, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True),
        ]
        if dropout > 0:
            layers.append(nn.Dropout2d(dropout))
        layers += [
            nn.Conv2d(out_channels, out_channels, kernel_size=3, padding=1, groups=out_channels, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True),
            nn.Conv2d(out_channels, out_channels, kernel_size=1, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True),
        ]
        self.block = nn.Sequential(*layers)

    def forward(self, x):
        return self.block(x)


class DecoderStage(nn.Module):
    def __init__(self, in_channels, skip_channels, out_channels, dropout=0.0):
        super().__init__()
        self.fuse = DepthwiseSeparableConv(in_channels + skip_channels, out_channels, dropout=dropout)

    def forward(self, x, skip=None, target_size=None):
        if target_size is not None:
            x = F.interpolate(x, size=target_size, mode="bilinear", align_corners=False)
        if skip is not None:
            x = torch.cat([x, skip], dim=1)
        return self.fuse(x)


class MobileNetV3Encoder(nn.Module):
    def __init__(self, pretrained=True):
        super().__init__()
        weights = MobileNet_V3_Large_Weights.DEFAULT if pretrained else None
        backbone = mobilenet_v3_large(weights=weights)
        self.features = backbone.features
        self.tap_points = {0, 2, 4, 7, 13, 16}

    def forward(self, x):
        feats = []
        for idx, layer in enumerate(self.features):
            x = layer(x)
            if idx in self.tap_points:
                feats.append(x)
        return feats


class MobileNetV3UNet(nn.Module):
    def __init__(self, num_classes, in_channels=3, pretrained_encoder=True, dropout=0.0):
        super().__init__()
        self.input_adapter = nn.Identity() if in_channels == 3 else nn.Conv2d(in_channels, 3, kernel_size=1, bias=False)
        self.encoder = MobileNetV3Encoder(pretrained=pretrained_encoder)

        self.decoder0 = DecoderStage(960, 160, 160, dropout=dropout)
        self.decoder1 = DecoderStage(160, 80, 80, dropout=dropout)
        self.decoder2 = DecoderStage(80, 40, 40, dropout=dropout)
        self.decoder3 = DecoderStage(40, 24, 24, dropout=dropout)
        self.decoder4 = DecoderStage(24, 16, 16, dropout=dropout)

        self.classifier = nn.Conv2d(16, num_classes, kernel_size=1)

    def forward(self, x):
        input_size = x.shape[-2:]
        x = self.input_adapter(x)

        feats = self.encoder(x)
        if len(feats) != 6:
            raise RuntimeError(f"Expected 6 feature maps, got {len(feats)}")

        skip1, skip2, skip3, skip4, skip5, bottleneck = feats

        x = self.decoder0(bottleneck, skip5, target_size=skip5.shape[-2:])
        x = self.decoder1(x, skip4, target_size=skip4.shape[-2:])
        x = self.decoder2(x, skip3, target_size=skip3.shape[-2:])
        x = self.decoder3(x, skip2, target_size=skip2.shape[-2:])
        x = self.decoder4(x, skip1, target_size=skip1.shape[-2:])

        x = F.interpolate(x, size=input_size, mode="bilinear", align_corners=False)
        return self.classifier(x)
```

```python
def train_one_epoch(model, loader, optimizer, criterion, device):
    model.train()
    running_loss = 0.0

    for images, masks in tqdm(loader):
        images = images.to(device)
        masks = masks.to(device)

        optimizer.zero_grad()
        logits = model(images)
        loss = criterion(logits, masks.long())
        loss.backward()
        optimizer.step()

        running_loss += loss.item()

    return running_loss / len(loader)


@torch.no_grad()
def evaluate(model, loader, criterion, device):
    model.eval()
    running_loss = 0.0

    for images, masks in loader:
        images = images.to(device)
        masks = masks.to(device)

        logits = model(images)
        loss = criterion(logits, masks.long())
        running_loss += loss.item()

    return running_loss / len(loader)
```

```python
epochs = 10
best_val = float("inf")

for epoch in range(epochs):
    train_loss = train_one_epoch(model, train_loader, optimizer, criterion, device)
    val_loss = evaluate(model, val_loader, criterion, device)

    print(f"Epoch {epoch + 1}/{epochs} - train_loss: {train_loss:.4f} - val_loss: {val_loss:.4f}")

    if val_loss < best_val:
        best_val = val_loss
        torch.save(model.state_dict(), "/content/best_teeth_unet_mobilenetv3.pth")
        print("Best model saved.")
```
